```python
import sys
from pathlib import Path

BASE = Path("/workspace/tiny-llm-from-scratch")
sys.path.append(str(BASE))

print(BASE)
```

```python
import json
import torch

from tokenizer.char_tokenizer import CharTokenizer
from model.tiny_transformer import TinyGPT, GPTConfig
```

```python
CHECKPOINT = BASE / "checkpoints/tinyllm_100k_char_best.pt"

checkpoint = torch.load(
    CHECKPOINT,
    map_location="cpu"
)

checkpoint.keys()
```

```python
checkpoint["config"]
```

```python
tokenizer = CharTokenizer.load(
    BASE / "tokenizer/tokenizer_char.json"
)

print(tokenizer.vocab_size)
```

```python
cfg = GPTConfig(**checkpoint["gpt_config"])

model = TinyGPT(cfg)

model.load_state_dict(checkpoint["model_state"])

model.eval()

print("Loaded successfully.")
```

```python
print("Step :", checkpoint["step"])
print("Parameters :", checkpoint["param_count"])
print("Best Loss :", checkpoint["best_val_loss"])
```

```python
prompt = "Hello"

ids = tokenizer.encode(prompt)

x = torch.tensor([ids])
```

```python
with torch.no_grad():

    output = model.generate(
        x,
        max_new_tokens=200,
        temperature=0.8,
        top_k=20
    )

generated = tokenizer.decode(
    output[0].tolist()
)

print(generated)
```

```python
prompts = [

    "Python",

    "Docker",

    "AI",

    "Machine Learning",

    "HTML",

    "SQL",

    "Once upon a time",

    "Assalamu Alaikum"

]

for prompt in prompts:

    ids = tokenizer.encode(prompt)

    x = torch.tensor([ids])

    output = model.generate(
        x,
        max_new_tokens=120
    )

    print("=" * 60)
    print(tokenizer.decode(output[0].tolist()))
```

```python
with torch.inference_mode():
    output = model.generate(
        x,
        max_new_tokens=200,
        temperature=0.8,
        top_k=20
    )
```
