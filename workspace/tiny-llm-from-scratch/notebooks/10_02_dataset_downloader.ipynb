```python
from pathlib import Path
import json
import shutil

from datasets import load_dataset
```

```python
BASE = Path("/workspace/tiny-llm-from-scratch")

RAW = BASE/"datasets"/"raw"

print(BASE)
```

```python
DATASETS = {
    "claude_mythos": {
        "enabled": True,
        "source": "huggingface",
        "dataset_id": "WithinUsAI/claude_mythos_distilled_25k",
        "split": "train",
        "save_as": "claude_mythos"
    }
}
```

```python
def download_huggingface(name, info):

    print(f"Downloading {name}")

    dataset = load_dataset(info["dataset_id"])

    return dataset
```

```python
dataset = download_huggingface(

    "claude_mythos",

    DATASETS["claude_mythos"]

)
```
