```python
import sys
import time
from pathlib import Path

import torch

BASE = Path("/workspace/tiny-llm-from-scratch")
sys.path.append(str(BASE))
```

```python
from tokenizer.char_tokenizer import CharTokenizer
from model.tiny_transformer import TinyGPT, GPTConfig
```

```python
checkpoint = torch.load(
    BASE/"checkpoints/tinyllm_100k_char_best.pt",
    map_location="cpu"
)

tokenizer = CharTokenizer.load(
    BASE/"tokenizer/tokenizer_char.json"
)

config = GPTConfig(**checkpoint["gpt_config"])

model = TinyGPT(config)

model.load_state_dict(checkpoint["model_state"])

model.eval()

print("Model Loaded Successfully")
```

```python
prompts = [

    "Python",

    "Docker",

    "Artificial Intelligence",

    "Machine Learning",

    "SQL",

    "FastAPI",

    "Once upon a time",

    "Assalamu Alaikum",

    "Transformer",

    "Hello"

]
```

```python
total_tokens = 0

start = time.time()

for prompt in prompts:

    ids = tokenizer.encode(prompt)

    x = torch.tensor([ids])

    output = model.generate(
        x,
        max_new_tokens=120,
        temperature=0.8,
        top_k=20
    )

    total_tokens += len(output[0]) - len(ids)

elapsed = time.time() - start

print("Generated Tokens :", total_tokens)

print("Time :", elapsed)

print("Speed :", total_tokens/elapsed,"tokens/sec")
```

```python
for prompt in prompts:

    ids = tokenizer.encode(prompt)

    x = torch.tensor([ids])

    output = model.generate(
        x,
        max_new_tokens=120,
        temperature=0.8,
        top_k=20
    )

    text = tokenizer.decode(output[0].tolist())

    print("="*70)
    print("PROMPT:")
    print(prompt)

    print()

    print("OUTPUT:")
    print(text)
```

```python
temperatures = [

    0.5,

    0.8,

    1.0,

    1.3

]

prompt = "Python"

for temp in temperatures:

    ids = tokenizer.encode(prompt)

    x = torch.tensor([ids])

    output = model.generate(
        x,
        max_new_tokens=120,
        temperature=temp,
        top_k=20
    )

    print("="*60)

    print("Temperature:", temp)

    print()

    print(tokenizer.decode(output[0].tolist()))
```

```python
topks = [

    5,

    10,

    20,

    40,

    None

]

prompt = "Machine Learning"

for k in topks:

    ids = tokenizer.encode(prompt)

    x = torch.tensor([ids])

    output = model.generate(
        x,
        max_new_tokens=120,
        temperature=0.8,
        top_k=k
    )

    print("="*60)

    print("Top-K:", k)

    print()

    print(tokenizer.decode(output[0].tolist()))
```
