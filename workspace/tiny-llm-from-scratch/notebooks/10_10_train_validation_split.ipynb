```python
import json
from pathlib import Path
```

```python
BASE = Path("/workspace/tiny-llm-from-scratch")

INPUT_FILE = (
    BASE /
    "datasets/processed/shuffled/shuffled_dataset.json"
)

OUTPUT_DIR = (
    BASE /
    "datasets/processed/split"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)
```

```python
with open(INPUT_FILE, "r", encoding="utf-8") as f:
    dataset = json.load(f)

print("Total Samples:", len(dataset))
```

```python
TRAIN_RATIO = 0.90
VAL_RATIO = 0.05
TEST_RATIO = 0.05

assert TRAIN_RATIO + VAL_RATIO + TEST_RATIO == 1.0
```

```python
total = len(dataset)

train_size = int(total * TRAIN_RATIO)

val_size = int(total * VAL_RATIO)

test_size = total - train_size - val_size

print("Train :", train_size)
print("Validation :", val_size)
print("Test :", test_size)
```

```python
train_data = dataset[:train_size]

validation_data = dataset[
    train_size:
    train_size + val_size
]

test_data = dataset[
    train_size + val_size:
]
```

```python
print(len(train_data))
print(len(validation_data))
print(len(test_data))

print()

print(
    len(train_data)
    +
    len(validation_data)
    +
    len(test_data)
)
```

```python
files = {

    "train.json": train_data,

    "validation.json": validation_data,

    "test.json": test_data

}

for filename, data in files.items():

    with open(
        OUTPUT_DIR / filename,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            data,
            f,
            indent=2,
            ensure_ascii=False
        )

print("Datasets Saved Successfully")
```

```python
for file in OUTPUT_DIR.iterdir():

    print(file.name)
```

```python
report = {

    "stage": "train_validation_test_split",

    "total_samples": total,

    "train_samples": len(train_data),

    "validation_samples": len(validation_data),

    "test_samples": len(test_data),

    "train_ratio": TRAIN_RATIO,

    "validation_ratio": VAL_RATIO,

    "test_ratio": TEST_RATIO

}

REPORT = (
    BASE /
    "reports/train_validation_split_report.json"
)

with open(
    REPORT,
    "w"
) as f:

    json.dump(
        report,
        f,
        indent=4
    )

print(REPORT)
```
