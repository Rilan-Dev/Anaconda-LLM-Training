```python
from pathlib import Path
import json
import re
from datasets import load_from_disk
```

```python
BASE = Path("/workspace/tiny-llm-from-scratch")

RAW_DATASET = BASE / "datasets/raw/claude_mythos"

OUTPUT_DIR = BASE / "datasets/processed/claude_mythos"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print(RAW_DATASET)
print(OUTPUT_DIR)
```

```python
dataset = load_from_disk(str(RAW_DATASET))

train = dataset["train"]

print(len(train))
```

```python
def clean_text(text):

    if text is None:
        return ""

    text = text.replace("\r", "")

    text = re.sub(r"\n{3,}", "\n\n", text)

    text = re.sub(r"[ \t]+", " ", text)

    text = text.strip()

    return text
```

```python
def clean_sample(sample):

    messages = sample["messages"]

    if len(messages) < 2:
        return None

    user = ""
    assistant = ""

    for msg in messages:

        if msg["role"] == "user":

            user = clean_text(msg["content"])

        elif msg["role"] == "assistant":

            assistant = clean_text(msg["content"])

    if user == "":
        return None

    if assistant == "":
        return None

    return {
        "user": user,
        "assistant": assistant,
        "category": sample["category"]
    }
```

```python
cleaned = []

removed = 0

for sample in train:

    item = clean_sample(sample)

    if item is None:

        removed += 1

    else:

        cleaned.append(item)

print("Clean Samples :", len(cleaned))

print("Removed :", removed)
```

```python
sample = cleaned[0]

print(sample["user"][:300])

print()

print(sample["assistant"][:500])
```

```python
output_file = OUTPUT_DIR / "clean_dataset.json"

with open(output_file, "w", encoding="utf-8") as f:

    json.dump(
        cleaned,
        f,
        ensure_ascii=False,
        indent=2
    )

print(output_file)
```

```python
with open(output_file, encoding="utf-8") as f:

    verify = json.load(f)

print(len(verify))

print()

print(verify[0].keys())
```
