```python
from pathlib import Path

BASE = Path("/workspace/tiny-llm-from-scratch")
DATA = BASE / "data"
DATA.mkdir(parents=True, exist_ok=True)

raw_text = """
Artificial intelligence is the science of making machines learn from data.
A language model learns to predict the next token in a sequence.
A tokenizer converts text into numbers.
A transformer is a neural network architecture used in modern language models.
Django is a Python web framework used to build web applications.
FastAPI is a Python framework used to build APIs.
Docker is used to package applications and run them inside containers.
A database stores structured information.
PostgreSQL is a relational database.
RAG means Retrieval Augmented Generation.
RAG helps a language model answer using external documents.
A vector database stores embeddings for semantic search.
Machine learning models improve by reducing loss during training.
The training loop includes forward pass, loss calculation, backpropagation, and optimizer step.
An LLM is trained using next-token prediction.
"""

# Repeat only for first tiny experiment
raw_text = raw_text * 500

(DATA / "raw.txt").write_text(raw_text, encoding="utf-8")

split_idx = int(len(raw_text) * 0.9)

train_text = raw_text[:split_idx]
val_text = raw_text[split_idx:]

(DATA / "train.txt").write_text(train_text, encoding="utf-8")
(DATA / "val.txt").write_text(val_text, encoding="utf-8")

print("Raw chars:", len(raw_text))
print("Train chars:", len(train_text))
print("Val chars:", len(val_text))
```
