```python
from pathlib import Path
import json

BASE = Path("/workspace/tiny-llm-from-scratch")
DATA = BASE / "data"
TOKENIZER_DIR = BASE / "tokenizer"

text = (DATA / "train.txt").read_text(encoding="utf-8")
chars = sorted(list(set(text)))

special_tokens = ["<UNK>"]
vocab = special_tokens + chars

stoi = {ch: i for i, ch in enumerate(vocab)}
itos = {i: ch for ch, i in stoi.items()}

payload = {
    "type": "char",
    "stoi": stoi,
    "itos": {str(k): v for k, v in itos.items()},
    "vocab_size": len(vocab)
}

(TOKENIZER_DIR / "tokenizer_char.json").write_text(
    json.dumps(payload, indent=2),
    encoding="utf-8"
)

print("Vocab size:", len(vocab))
print("Tokenizer saved:", TOKENIZER_DIR / "tokenizer_char.json")
```

```python
def encode(text):
    return [stoi.get(ch, stoi["<UNK>"]) for ch in text]

def decode(ids):
    return "".join([itos.get(int(i), "<UNK>") for i in ids])

sample = "Artificial intelligence is"
ids = encode(sample)

print(ids)
print(decode(ids))
```
