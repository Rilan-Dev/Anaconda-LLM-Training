```python
samples = dataset.head(3)

for i, sample in enumerate(samples, start=1):

    print("="*80)

    print(f"Sample {i}")

    print("-"*80)

    print("Instruction:\n")

    print(sample.instruction[:300])

    print()

    print("Response:\n")

    print(sample.response[:500])

    print()
```

```python
if results:

    print("="*80)

    print("Instruction")

    print("-"*80)

    print(results[0].instruction)

    print()

    print("="*80)

    print("Response")

    print("-"*80)

    print(results[0].response[:1000])
```
