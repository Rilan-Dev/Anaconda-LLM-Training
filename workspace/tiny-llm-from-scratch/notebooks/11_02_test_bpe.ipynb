```python
prompts = [

    "Python is awesome.",

    "Docker containers.",

    "Artificial Intelligence.",

    "Assalamu Alaikum.",

    "TinyGPT",

    "Deep Learning",

]
```

```python
for text in prompts:

    result = tokenizer.encode_with_tokens(text)

    print("=" * 60)

    print(text)

    print()

    print(result["tokens"])

    print(result["ids"])
```

```python
for text in prompts:

    ids = tokenizer.encode(text)

    decoded = tokenizer.decode(ids)

    print("=" * 60)

    print(text)

    print(decoded)

    print(text == decoded)
```
