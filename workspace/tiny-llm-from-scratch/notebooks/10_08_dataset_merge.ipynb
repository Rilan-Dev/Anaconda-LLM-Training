```python
from pathlib import Path
import json
```

```python
BASE = Path("/workspace/tiny-llm-from-scratch")

INPUT_DIR = BASE / "datasets/processed"

OUTPUT_DIR = INPUT_DIR / "merged"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

OUTPUT_FILE = OUTPUT_DIR / "merged_dataset.json"
```

```python
dataset_files = list(INPUT_DIR.glob("*/deduplicated_dataset.json"))

print("Datasets Found")

print("="*50)

for file in dataset_files:
    print(file)
```

```python
merged = []

dataset_info = []

for file in dataset_files:

    with open(file, encoding="utf-8") as f:

        data = json.load(f)

    dataset_name = file.parent.name

    dataset_info.append({

        "dataset": dataset_name,

        "samples": len(data)

    })

    for sample in data:

        sample["dataset"] = dataset_name

        merged.append(sample)

print("Merged Samples :", len(merged))
```

```python
print(merged[0])
```

```python
with open(
    OUTPUT_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(

        merged,

        f,

        indent=2,

        ensure_ascii=False

    )

print(OUTPUT_FILE)
```

```python
print("="*60)

print("Merged Dataset Report")

print("="*60)

total = 0

for info in dataset_info:

    print(f'{info["dataset"]:25} {info["samples"]:8}')

    total += info["samples"]

print("-"*60)

print(f'{"TOTAL":25} {total:8}')

print("="*60)
```

```python
with open(OUTPUT_FILE, encoding="utf-8") as f:

    verify = json.load(f)

print(len(verify))

print()

print(verify[0].keys())
```

```python
from collections import Counter

counter = Counter()

for sample in verify:

    counter[sample["dataset"]] += 1

print(counter)
```

```python
print("="*70)

print("DATASET MERGE COMPLETED")

print("="*70)

print("Datasets merged :", len(dataset_info))

print("Total samples   :", len(verify))

print("Output file")

print(OUTPUT_FILE)

print()

print("Ready for Phase 10.09")

print("="*70)
```
