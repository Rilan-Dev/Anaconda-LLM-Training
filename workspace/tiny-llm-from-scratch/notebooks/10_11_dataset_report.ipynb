```python
import json
from pathlib import Path
from collections import Counter
```

```python
BASE = Path("/workspace/tiny-llm-from-scratch")

TRAIN_FILE = (
    BASE /
    "datasets/processed/split/train.json"
)

REPORT_DIR = BASE / "reports"

REPORT_DIR.mkdir(
    parents=True,
    exist_ok=True
)
```

```python
with open(TRAIN_FILE, "r", encoding="utf-8") as f:
    dataset = json.load(f)

print("Training Samples:", len(dataset))
```

```python
instruction_lengths = []

response_lengths = []

categories = Counter()

sources = Counter()

empty_records = 0
```

```python
for sample in dataset:

    instruction = sample.get("instruction", "")
    response = sample.get("response", "")

    instruction_lengths.append(len(instruction))
    response_lengths.append(len(response))

    if not instruction.strip() or not response.strip():
        empty_records += 1

    categories[sample.get("category", "Unknown")] += 1
    sources[sample.get("source", "Unknown")] += 1
```

```python
report = {

    "total_samples": len(dataset),

    "empty_records": empty_records,

    "average_instruction_length":
        sum(instruction_lengths) / len(instruction_lengths),

    "average_response_length":
        sum(response_lengths) / len(response_lengths),

    "longest_instruction":
        max(instruction_lengths),

    "shortest_instruction":
        min(instruction_lengths),

    "longest_response":
        max(response_lengths),

    "shortest_response":
        min(response_lengths),

    "categories":
        dict(categories),

    "sources":
        dict(sources)

}

report
```

```python
REPORT_JSON = (
    REPORT_DIR /
    "dataset_quality_report.json"
)

with open(
    REPORT_JSON,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        report,
        f,
        indent=4,
        ensure_ascii=False
    )

print(REPORT_JSON)
```

```python
REPORT_TEXT = (
    REPORT_DIR /
    "dataset_statistics.txt"
)

with open(
    REPORT_TEXT,
    "w",
    encoding="utf-8"
) as f:

    f.write("TinyGPT Dataset Report\n")
    f.write("=" * 50 + "\n\n")

    for key, value in report.items():

        f.write(f"{key}\n")

        f.write(f"{value}\n\n")

print(REPORT_TEXT)
```

```python
print("=" * 60)

print("Dataset Quality Report")

print("=" * 60)

print("Samples:", report["total_samples"])

print("Empty Records:", report["empty_records"])

print()

print(
    "Average Instruction Length:",
    round(report["average_instruction_length"], 2)
)

print(
    "Average Response Length:",
    round(report["average_response_length"], 2)
)

print()

print(
    "Longest Instruction:",
    report["longest_instruction"]
)

print(
    "Longest Response:",
    report["longest_response"]
)

print()

print("Categories")

print("-" * 30)

for name, count in categories.items():

    print(f"{name:<30} {count}")

print()

print("Sources")

print("-" * 30)

for name, count in sources.items():

    print(f"{name:<40} {count}")

print("=" * 60)
```

```python
print("=" * 60)

print("Dataset Health Check")

print("=" * 60)

checks = {

    "No Empty Records":
        empty_records == 0,

    "Enough Training Samples":
        len(dataset) >= 10000,

    "Average Response > 100 chars":
        report["average_response_length"] > 100,

    "Average Instruction > 20 chars":
        report["average_instruction_length"] > 20

}

for name, result in checks.items():

    print(
        f"{name:<40}",
        "PASS" if result else "FAIL"
    )

print("=" * 60)
```
