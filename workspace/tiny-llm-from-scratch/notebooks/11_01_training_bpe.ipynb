```python
from pathlib import Path

PROJECT_ROOT = Path("/workspace/tiny-llm-from-scratch")

print(PROJECT_ROOT)
```

```python
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace
```

```python
import json

dataset_path = PROJECT_ROOT / "datasets/processed/merged/merged_dataset.json"

with open(dataset_path, encoding="utf-8") as f:
    dataset = json.load(f)

print(len(dataset))
```

```python
corpus = []

for sample in dataset:

    if sample["user"].strip():
        corpus.append(sample["dataset"])

    if sample["user"].strip():
        corpus.append(sample["dataset"])

print(len(corpus))
```

```python
training_file = PROJECT_ROOT / "datasets/processed/bpe_corpus.txt"

with open(training_file, "w", encoding="utf-8") as f:

    for line in corpus:

        f.write(line + "\n")

print(training_file)
```

```python
tokenizer = Tokenizer(BPE())

tokenizer.pre_tokenizer = Whitespace()
```

```python
trainer = BpeTrainer(

    vocab_size=8000,

    min_frequency=2,

    special_tokens=[

        "<PAD>",
        "<UNK>",
        "<BOS>",
        "<EOS>",

    ],
)
```

```python
tokenizer.train(

    [str(training_file)],

    trainer,
)
```

```python
save_path = PROJECT_ROOT / "tokenizer/bpe/tokenizer.json"

tokenizer.save(str(save_path))

print("Saved:", save_path)
```

```python
tokenizer = Tokenizer.from_file(str(save_path))
```

```python
text = "Artificial Intelligence is amazing."

encoded = tokenizer.encode(text)

print(encoded.tokens)

print(encoded.ids)
```

```python
print(

    tokenizer.decode(

        encoded.ids

    )
)
```
