```python
from pathlib import Path

import json

import re

import unicodedata
```

```python
BASE = Path("/workspace/tiny-llm-from-scratch")

INPUT = BASE / "datasets/processed/claude_mythos/clean_dataset.json"

OUTPUT = BASE / "datasets/processed/claude_mythos/normalized_dataset.json"

print(INPUT)
```

```python
with open(INPUT, encoding="utf-8") as f:

    dataset = json.load(f)

print(len(dataset))
```

```python
def normalize_text(text):

    if text is None:

        return ""

    # Unicode normalization
    text = unicodedata.normalize("NFKC", text)

    # Standardize line endings
    text = text.replace("\r\n", "\n")
    text = text.replace("\r", "\n")

    # Replace smart quotes
    text = text.replace("“", '"')
    text = text.replace("”", '"')
    text = text.replace("‘", "'")
    text = text.replace("’", "'")

    # Replace tabs with spaces
    text = text.replace("\t", " ")

    # Remove multiple spaces
    text = re.sub(r"[ ]{2,}", " ", text)

    # Remove excessive blank lines
    text = re.sub(r"\n{3,}", "\n\n", text)

    # Remove trailing spaces
    text = "\n".join(line.rstrip() for line in text.splitlines())

    return text.strip()
```

```python
normalized = []

for sample in dataset:

    normalized.append({

        "user": normalize_text(sample["user"]),

        "assistant": normalize_text(sample["assistant"]),

        "category": sample["category"]

    })

print(len(normalized))
```

```python
sample = normalized[0]

print("="*60)

print(sample["user"])

print()

print("-"*60)

print()

print(sample["assistant"][:1000])
```

```python
with open(OUTPUT, "w", encoding="utf-8") as f:

    json.dump(

        normalized,

        f,

        ensure_ascii=False,

        indent=2

    )

print(OUTPUT)
```
