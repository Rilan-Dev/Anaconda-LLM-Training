```python
from pathlib import Path
import sys

BASE = Path("/workspace/tiny-llm-from-scratch")

sys.path.append(str(BASE))

print("=" * 60)
print("TinyGPT Interactive Chat")
print("=" * 60)
print("Project :", BASE)
```

```python
import json
import time

import torch

from tokenizer.char_tokenizer import CharTokenizer
from model.tiny_transformer import TinyGPT, GPTConfig
```

```python
TOKENIZER_PATH = BASE / "tokenizer/tokenizer_char.json"

tokenizer = CharTokenizer.load(TOKENIZER_PATH)

print("Tokenizer Loaded")
print("Vocabulary Size :", tokenizer.vocab_size)
```

```python
CHECKPOINT = BASE / "checkpoints/tinyllm_100k_char_best.pt"

checkpoint = torch.load(
    CHECKPOINT,
    map_location="cpu"
)

print("Checkpoint Loaded")

print("Training Step :", checkpoint["step"])

print("Best Loss :", checkpoint["best_val_loss"])
```

```python
cfg = GPTConfig(**checkpoint["gpt_config"])

model = TinyGPT(cfg)

model.load_state_dict(checkpoint["model_state"])

model.eval()

print("Model Restored Successfully")
```

```python
def generate_text(
    prompt,
    max_new_tokens=120,
    temperature=0.8,
    top_k=20,
):
    ids = tokenizer.encode(prompt)

    x = torch.tensor([ids])

    start = time.time()

    with torch.inference_mode():

        output = model.generate(
            x,
            max_new_tokens=max_new_tokens,
            temperature=temperature,
            top_k=top_k,
        )

    elapsed = time.time() - start

    text = tokenizer.decode(output[0].tolist())

    return text, elapsed
```

```python
def chat(
    prompt,
    temperature=0.8,
    top_k=20,
    max_new_tokens=120,
):
    print()

    print("=" * 60)

    print("You:")

    print(prompt)

    print()

    response, elapsed = generate_text(
        prompt,
        max_new_tokens=max_new_tokens,
        temperature=temperature,
        top_k=top_k,
    )

    print("TinyGPT:")

    print(response)

    print()

    print(f"Generation Time : {elapsed:.2f} sec")

    print("=" * 60)
```

```python
print("Type 'exit' to stop.")

while True:

    prompt = input("\nYou: ")

    if prompt.lower() in ["exit", "quit"]:

        print("\nTinyGPT: Goodbye!")

        break

    chat(prompt)
```

```python
TEMPERATURE = 0.8

TOP_K = 20

MAX_NEW_TOKENS = 120

print("Current Settings")

print("-" * 30)

print("Temperature :", TEMPERATURE)

print("Top-K :", TOP_K)

print("Max Tokens :", MAX_NEW_TOKENS)
```

```python
prompt = "Python"

response, elapsed = generate_text(
    prompt,
    temperature=TEMPERATURE,
    top_k=TOP_K,
    max_new_tokens=MAX_NEW_TOKENS,
)

print("=" * 50)

print("Characters Generated")

print(len(response))

print()

print("Generation Time")

print(f"{elapsed:.3f} seconds")

print()

print("Characters / Second")

print(len(response) / elapsed)

print("=" * 50)
```

```python
prompts = [
    "Python",
    "Docker",
    "AI",
    "Machine Learning",
    "SQL",
    "FastAPI",
    "Transformer",
]

for prompt in prompts:

    chat(prompt)
```
