```python
import json
import random
from pathlib import Path
```

```python
BASE = Path("/workspace/tiny-llm-from-scratch")

INPUT_FILE = (
    BASE /
    "datasets/processed/merged/merged_dataset.json"
)

OUTPUT_FILE = (
    BASE /
    "datasets/processed/shuffled/shuffled_dataset.json"
)

REPORT_FILE = (
    BASE /
    "reports/dataset_shuffle_report.json"
)
```

```python
with open(INPUT_FILE, "r", encoding="utf-8") as f:
    records = json.load(f)

print("Loaded Samples:", len(records))
```

```python
print(type(records))
print(records[0].keys())
```

```python
SEED = 42

random.seed(SEED)

random.shuffle(records)

print("Dataset shuffled successfully.")
```

```python
OUTPUT_FILE.parent.mkdir(
    parents=True,
    exist_ok=True
)

with open(
    OUTPUT_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        records,
        f,
        indent=2,
        ensure_ascii=False
    )

print("Saved:", OUTPUT_FILE)
```
