```python
from pathlib import Path
import json
import platform
import sys
```

```python
BASE = Path("/workspace/tiny-llm-from-scratch")

print(BASE)
```

```python
DIRECTORIES = [

    "dataset_pipeline",

    "datasets/raw",
    "datasets/cleaned",
    "datasets/normalized",
    "datasets/merged",
    "datasets/final",
    "datasets/reports",
    "datasets/cache",

    "configs"

]
```

```python
for directory in DIRECTORIES:

    path = BASE / directory

    path.mkdir(parents=True, exist_ok=True)

    print("✓", path)
```

```python
config_path = BASE / "configs" / "dataset_pipeline.json"

with open(config_path, "w", encoding="utf-8") as f:

    json.dump(
        PIPELINE_CONFIG,
        f,
        indent=4
    )

print("Saved:", config_path)
```
